## Logistice regression practical implementation

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
x,y = make_classification(n_samples=1000,n_features=10,n_classes =2,random_state=15)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_model = LogisticRegression()
logistic_model.fit(x_train,y_train)
```

```python
y_pred = logistic_model.predict(x_test)
```

```python
print(y_pred)
```

```python
plt.scatter(x_test[:,0],x_test[:,1],c=y_pred)

```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
score = accuracy_score(y_test,y_pred)
print("Accuracy:", score)
classification_rep = classification_report(y_test,y_pred)
print("Classification Report:\n", classification_rep)
confusion_mat = confusion_matrix(y_test,y_pred)
print("Confusion Matrix:\n", confusion_mat)
```

## Hyperparameter Tuning And Cross Validation


## Grid SearchCV


```python
model = LogisticRegression()
penatly = ['l1','l2','elasticnet']
c_values=[100,10,1.0,0.1,0.01]
solver=['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']
```

```python
params = dict(penalty=penatly,C=c_values,solver=solver)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid = GridSearchCV(estimator=model,param_grid=params,cv=5,scoring='accuracy',n_jobs=-1)
```

```python
grid
```

```python
grid.fit(x_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred  = grid.predict(x_test)
```

```python
score = accuracy_score(y_test,y_pred)
print("Accuracy:", score)
classification_rep = classification_report(y_test,y_pred)
print("Classification Report:\n", classification_rep)
confusion_mat = confusion_matrix(y_test,y_pred)
print("Confusion Matrix:\n", confusion_mat)

```

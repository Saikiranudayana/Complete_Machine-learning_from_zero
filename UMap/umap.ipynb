```python
from sklearn.datasets import fetch_openml

##Load data
mnist_data = fetch_openml('mnist_784',version=1,as_frame=False)

```

```python
x = mnist_data.data
y = mnist_data.target
```

```python
print(x.shape)
```

```python
print(y.shape)
```

```python
import umap

reducer = umap.UMAP(n_neighbors = 15)


x_umap = reducer.fit_transform(x)
print(x_umap.shape)
```

784- > 2 features

## visulaize


```python
import matplotlib.pyplot as plt 

plt.figure(figsize = (12,8))


scatter = plt.scatter(
    x_umap[:,0],
    x_umap[:,1],
    c=y.astype(int),
    cmap = 'tab10',
    s= 5
    
)

plt.colorbar(scatter)
plt.title("UMap projection of Mnist")
plt.xlabel("UMap compenet 1")
plt.ylabel("UMap compenent 2")


plt.show()
```
